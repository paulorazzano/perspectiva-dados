```python
distribuicoes["Normal"]["params"] = {"loc": 0, "scale": 1}
distribuicoes["Exponencial"]["params"] = {"scale": 5}
distribuicoes["Uniforme"]["params"] = {"low": 0, "high": 5}
distribuicoes["Poisson"]["params"] = {"lam": 3}

for nome in distribuicoes:

    if nome == "Normal":
        media_teorica = distribuicoes[nome]["params"]["loc"]
        variancia_teorica = distribuicoes[nome]["params"]["scale"] ** 2
        dp_teorico = distribuicoes[nome]["params"]["scale"]

    elif nome == "Exponencial":
        media_teorica = distribuicoes[nome]["params"]["scale"]
        variancia_teorica = distribuicoes[nome]["params"]["scale"] ** 2
        dp_teorico = distribuicoes[nome]["params"]["scale"]

    elif nome == "Uniforme":
        a = distribuicoes[nome]["params"]["low"]
        b = distribuicoes[nome]["params"]["high"]

        media_teorica = (a + b) / 2
        variancia_teorica = ((b - a) ** 2) / 12
        dp_teorico = np.sqrt(variancia_teorica)

    elif nome == "Poisson":
        media_teorica = distribuicoes[nome]["params"]["lam"]
        variancia_teorica = distribuicoes[nome]["params"]["lam"]
        dp_teorico = np.sqrt(distribuicoes[nome]["params"]["lam"])

    distribuicoes[nome]["estatisticas"]["media_teorica"] = media_teorica
    distribuicoes[nome]["estatisticas"]["variancia_teorica"] = variancia_teorica
    distribuicoes[nome]["estatisticas"]["dp_teorico"] = dp_teorico

nomes_distribuicoes = []
medias_amostra = []
medianas = []
variancias_amostra = []
desvios_padrao_amostra = []
minimos = []
maximos = []
quartis_1 = []
quartis_2 = []
quartis_3 = []
medias_teoricas = []
variancias_teoricas = []
desvios_padrao_teoricos = []

for nome in distribuicoes:

    nomes_distribuicoes.append(nome)

    medias_amostra.append(distribuicoes[nome]["estatisticas"]["media"])
    medianas.append(distribuicoes[nome]["estatisticas"]["mediana"])
    variancias_amostra.append(distribuicoes[nome]["estatisticas"]["var"])
    desvios_padrao_amostra.append(distribuicoes[nome]["estatisticas"]["dp"])

    minimos.append(distribuicoes[nome]["estatisticas"]["min"])
    maximos.append(distribuicoes[nome]["estatisticas"]["max"])
    quartis_1.append(distribuicoes[nome]["estatisticas"]["q1"])
    quartis_2.append(distribuicoes[nome]["estatisticas"]["q2"])
    quartis_3.append(distribuicoes[nome]["estatisticas"]["q3"])

    medias_teoricas.append(distribuicoes[nome]["estatisticas"]["media_teorica"])
    variancias_teoricas.append(distribuicoes[nome]["estatisticas"]["variancia_teorica"])
    desvios_padrao_teoricos.append(distribuicoes[nome]["estatisticas"]["dp_teorico"])

df_resultados = pd.DataFrame({
    "Distribuicao": nomes_distribuicoes,
    "Media_Amostra": medias_amostra,
    "Media_Teorica": medias_teoricas,
    "Mediana": medianas,
    "Variancia_Amostra": variancias_amostra,
    "Variancia_Teorica": variancias_teoricas,
    "Desvio_Padrao_Amostra": desvios_padrao_amostra,
    "Desvio_Padrao_Teorico": desvios_padrao_teoricos,
    "Minimo": minimos,
    "Maximo": maximos,
    "Q1": quartis_1,
    "Q2": quartis_2,
    "Q3": quartis_3
})

print(df_resultados)
```

```python
def calcular_produto_e_distancia(vetor1, vetor2):

    produto_escalar = np.dot(vetor1, vetor2)

    distancia_euclidiana = np.linalg.norm(vetor1 - vetor2)

    return produto_escalar, distancia_euclidiana

vetor_a = np.array([1, 2, 3])
vetor_b = np.array([4, 5, 6])

produto, distancia = calcular_produto_e_distancia(vetor_a, vetor_b)

print("Vetor 1:", vetor_a)
print("Vetor 2:", vetor_b)
print("Produto escalar:", produto)
print("Distância euclidiana:", distancia)
```

```python
M = np.random.randint(-5, 6, (5, 5))

print("Matriz M:")
print(M)

B = (1/2) * (M + M.T)

print("\nMatriz simétrica B:")
print(B)

autovalores, autovetores = np.linalg.eig(B)

print("\nAutovalores:")
print(autovalores)

print("\nAutovetores:")
print(autovetores)

print("\nVerificação da relação Bv = lambda v:\n")

for i in range(len(autovalores)):
    lambda_i = autovalores[i]
    v_i = autovetores[:, i]

    lado_esquerdo = np.dot(B, v_i)
    lado_direito = lambda_i * v_i

    print(f"Autovalor {i+1}: {lambda_i}")
    print("Bv =")
    print(lado_esquerdo)
    print("lambda * v =")
    print(lado_direito)

    if np.allclose(lado_esquerdo, lado_direito):
        print("Verificação certa: Bv = lambda*v\n")
    else:
        print("Verificação errada\n")
```

```python
A = np.random.randint(-5, 6, (3, 3))

y = np.random.randint(-5, 6, 3)

print("Matriz A:")
print(A)

print("\nVetor y:")
print(y)

detA = np.linalg.det(A)

print("\nDeterminante de A:")
print(detA)

if not np.isclose(detA, 0):
    x = np.linalg.solve(A, y)

    print("\nSolução x do sistema Ax = y:")
    print(x)

    Ax = np.dot(A, x)

    print("\nProduto Ax:")
    print(Ax)

    diferenca = Ax - y

    print("\nDiferença Ax - y:")
    print(diferenca)

    norma_quadrado = np.linalg.norm(diferenca) ** 2

    print("\nValor de ||Ax - y||^2:")
    print(norma_quadrado)

else:
    print("\nA matriz gerada não é invertível, então não dá para resolver o sistema com np.linalg.solve.")
```

```python
np.random.seed(10)

dados = np.random.normal(loc=0, scale=1, size=(100, 2))

print("Primeiros vetores gerados:")
print(dados[:5])

S = np.cov(dados, rowvar=False)

print("\nMatriz de covariância S:")
print(S)

autovalores, autovetores = np.linalg.eig(S)

print("\nAutovalores:")
print(autovalores)

print("\nAutovetores:")
print(autovetores)

indices = np.argsort(autovalores)[::-1]

autovalores = autovalores[indices]
autovetores = autovetores[:, indices]

print("\nAutovalores em ordem decrescente:")
print(autovalores)

print("\nAutovetores reordenados:")
print(autovetores)

media = np.mean(dados, axis=0)

print("\nMédia dos dados:")
print(media)

plt.figure(figsize=(8, 8))

plt.scatter(dados[:, 0], dados[:, 1], alpha=0.6, label="Dados")

plt.scatter(media[0], media[1], marker="x", s=100, label="Média")

for i in range(2):
    vetor = autovetores[:, i]
    comprimento = np.sqrt(autovalores[i]) * 2

    plt.arrow(
        media[0], media[1],
        vetor[0] * comprimento, vetor[1] * comprimento,
        head_width=0.1,
        length_includes_head=True
    )

plt.title("Direções principais da variabilidade dos dados")
plt.xlabel("Variável 1")
plt.ylabel("Variável 2")
plt.axhline(0)
plt.axvline(0)
plt.legend()
plt.grid(True)
plt.axis("equal")
plt.show()
```
