```python
import requests
import pandas as pd
import unicodedata
from bs4 import BeautifulSoup
from collections import Counter
from tqdm import tqdm
```

```python
# URL da vítima
url = "https://lbf.com.br/noticias/"

# Cabeçalho para simular um navegador
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}
```

```python
response = requests.get(url, headers=headers)
response
# <Response [200]> -> requisição foi bem-sucedida
```

```python
response.text
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
# Exibe o html como texto
print(soup.get_text())
```

```python
# Localizando o bloco com o id "noticiasInternas"
bloco_noticias = soup.find("div", id="noticiasInternas")

bloco_noticias
```

```python
links = [a['href'] for a in bloco_noticias.find_all('a', href=True)]

# Exibindo os links extraídos
for link in links:
    print(link)
```

```python
links_noticias = set()

for pagina in tqdm(range(1, 331)):
    url = "https://lbf.com.br/noticias/" if pagina == 1 else f"https://lbf.com.br/noticias/page/{pagina}/"
    html = requests.get(url).text
    soup = BeautifulSoup(html, "html.parser")

    for a in soup.find_all("a", href=True):
        link = a["href"]
        if "/noticias/" in link and "/noticias/page/" not in link:
            links_noticias.add(link)

for link in links_noticias:
    print(link)
```

```python
print(len(links_noticias))
```

```python
times = {
    "ADRM Maringá": ["ADRM Maringá", "Maringá"],
    "Cerrado BRB": ["Cerrado BRB", "Cerrado"],
    "Pontz São José Basketball": ["Pontz São José Basketball", "São José", "Sao Jose"],
    "Salvador Basketball": ["Salvador Basketball", "Salvador"],
    "Sampaio Basquete": ["Sampaio Basquete", "Sampaio"],
    "Santo André": ["Santo André", "Santo Andre"],
    "SESI Araraquara Basquete": ["SESI Araraquara Basquete", "SESI Araraquara", "Araraquara"],
    "Sodiê Mesquita": ["Sodiê Mesquita", "Sodie Mesquita", "Mesquita"],
    "Sport/Uninassau/Inst. Todos": ["Sport/Uninassau/Inst. Todos", "Sport", "Uninassau"],
    "Unimed Campinas Basquete": ["Unimed Campinas Basquete", "Unimed Campinas", "Campinas"]
}

def normalizar_texto(texto):
    texto = unicodedata.normalize("NFD", texto)
    texto = "".join(c for c in texto if unicodedata.category(c) != "Mn")
    return texto.lower()

dados = []
sessao = requests.Session()

for link in tqdm(list(links_noticias)[:1000], desc="Lendo notícias"):
    try:
        html = sessao.get(link, timeout=10).text
        soup = BeautifulSoup(html, "html.parser")

        titulo = link.rstrip("/").split("/")[-1].replace("-", " ")

        texto = " ".join(p.get_text(" ", strip=True) for p in soup.find_all("p"))
        texto_normalizado = normalizar_texto(texto)

        contador = Counter()

        for time, nomes in times.items():
            qtd = 0
            for nome in nomes:
                qtd += texto_normalizado.count(normalizar_texto(nome))
            if qtd > 0:
                contador[time] = qtd

        qtd_palavras = len(texto.split())
        qtd_times_citados = len(contador)

        if contador:
            time_mais_citado, qtd_citacoes = contador.most_common(1)[0]
        else:
            time_mais_citado = "Nenhum time identificado"
            qtd_citacoes = 0

        dados.append({
            "titulo_noticia": titulo,
            "time_mais_citado": time_mais_citado,
            "qtd_citacoes_time_mais_citado": qtd_citacoes,
            "qtd_total_palavras": qtd_palavras,
            "qtd_times_citados": qtd_times_citados,
            "link": link
        })

    except:
        pass

df = pd.DataFrame(dados)
print(df)
```

```python
df.to_csv("noticias_lbf.csv", index=False, encoding="utf-8-sig")
```
